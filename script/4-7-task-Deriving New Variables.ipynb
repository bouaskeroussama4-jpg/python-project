#### If-Statements with User-Defined Functions

```python
def price_label(row):

  if row['prices'] <= 5:
    return 'Low-range product'
  elif (row['prices'] > 5) and (row['prices'] <= 15):
    return 'Mid-range product'
  elif row['prices'] > 15:
    return 'High range'
  else: return 'Not enough data'
```

#### processing to get result

```python
result_2 = []

for value in ords_prods_merge["order_dow"]:
    if value >= 0 and value <= 1:
        result_2.append("Busiest days")
    elif value >= 3 and value <= 4:
        result_2.append("Least busy")
    else:
        result_2.append("Regularly busy")
```

```python
# creation of column
ords_prods_merge['Busiest days'] = result_2
```

# 4/The senior technical officer at Instacart wants you to identify the busiest hours of the day. Rather than by hour, they want periods of time labeled “Most orders,” “Average orders,” and “Fewest orders.” Create a new column containing these labels called “busiest_period_of_day.”

## search for the partition method

```python
ords_prods_merge['order_hour_of_day'].value_counts(dropna = False)
```

### processing

```python
# Fewest orders : 0–7
ords_prods_merge.loc[(ords_prods_merge['order_hour_of_day'] >= 0) & (ords_prods_merge['order_hour_of_day'] <= 7),
       'busiest_period_of_day'] = 'Fewest orders'


```

```python
# Most orders : 8–15
ords_prods_merge.loc[(ords_prods_merge['order_hour_of_day'] >= 8) & (ords_prods_merge['order_hour_of_day'] <= 15),
       'busiest_period_of_day'] = 'Most orders'
```

```python
# Average orders : 16–23
ords_prods_merge.loc[(ords_prods_merge['order_hour_of_day'] >= 16) & (ords_prods_merge['order_hour_of_day'] <= 23),
       'busiest_period_of_day'] = 'Average orders'
```
