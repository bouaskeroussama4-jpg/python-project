## 6/Create a spending flag for each user based on the average price across all their orders

##### creating mean price colomn

```python
Deriving_New_Variables['mean_price'] = Deriving_New_Variables.groupby(['user_id'])['prices'].transform(np.mean)
```

##### creating spending flag colomn

```python
Deriving_New_Variables.loc[Deriving_New_Variables['mean_price'] >= 10, 'spending_flag'] = 'High spender'
Deriving_New_Variables.loc[Deriving_New_Variables['mean_price'] < 10, 'spending_flag'] = 'Low spender'
```

##### checking results

```python
Deriving_New_Variables[['user_id', 'mean_price', 'spending_flag']].head(10)
```

##### second check

```python
Deriving_New_Variables['spending_flag'].value_counts()
```

## 7/Create an order frequency flag that marks the regularity of a user’s ordering behavior according to the median in the “days_since_prior_order” column.

##### creating median for days since last order

```python
Deriving_New_Variables['median_days_since_last_orders'] = Deriving_New_Variables.groupby(['user_id'])['days_since_prior_order'].transform(np.median)
```

##### creating order frequency flag

```python
Deriving_New_Variables.loc[Deriving_New_Variables['median_days_since_last_orders'] > 20, 'order_frequency_flag'] = 'Non-frequent customer'
Deriving_New_Variables.loc[(Deriving_New_Variables['median_days_since_last_orders'] <= 20) & (Deriving_New_Variables['median_days_since_last_orders'] > 10), 'order_frequency_flag'] = 'Regular customer'
Deriving_New_Variables.loc[Deriving_New_Variables['median_days_since_last_orders'] <= 10, 'order_frequency_flag'] = 'Frequent customer'
```

##### checking result

```python
Deriving_New_Variables[['user_id', 'median_days_since_last_orders', 'order_frequency_flag']].head(10)
```

##### second check

```python
Deriving_New_Variables['order_frequency_flag'].value_counts()
```
